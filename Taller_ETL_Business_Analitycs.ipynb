# **Contexto del Proceso ETL**

Este proceso se utiliza para obtener información desde una o varias fuentes, modificarla para mejorar su calidad y estructura, y posteriormente almacenarla en un destino donde pueda ser utilizada para análisis o procesos empresariales.

El proceso se divide en tres etapas principales:

- **Extracción**

Consiste en obtener los datos desde una fuente de origen. Estas fuentes pueden ser archivos, bases de datos, sistemas empresariales, APIs u otros medios de almacenamiento.

- **Transformación**

En esta etapa los datos obtenidos son revisados y procesados para mejorar su calidad y facilitar su utilización. Entre las actividades que pueden realizarse se encuentran la identificación de valores faltantes, eliminación de registros duplicados, modificación de tipos de datos, filtrado de información y creación de nuevas variables.

- **Carga**

La carga consiste en almacenar los datos después de haber sido transformados en un destino adecuado para su utilización. En este proyecto, los datos procesados serán almacenados en un archivo CSV que podrá utilizarse posteriormente para procesos de análisis.

# **Transformación y limpieza de los datos**

En esta etapa se aplicarán diferentes procedimientos para mejorar la calidad y estructura de los datos. Las transformaciones se realizarán de manera progresiva, teniendo en cuenta los problemas identificados durante el análisis anterior.

***

**Eliminación de registros duplicados**

Durante el análisis inicial se identificaron 5.268 registros duplicados.

Los registros duplicados contienen exactamente los mismos valores en todas sus columnas. Mantenerlos podría provocar que determinadas operaciones fueran contabilizadas más de una vez y afectar cálculos posteriores.

Por esta razón, se eliminarán los registros completamente duplicados, conservando únicamente una copia de cada registro.

```python
# Guardar la cantidad de registros antes de eliminar duplicados
registros_antes = len(df_retail)

# Eliminar los registros que están completamente duplicados
df_retail = df_retail.drop_duplicates()

# Guardar la cantidad de registros después de eliminar duplicados
registros_despues = len(df_retail)

# Calcular cuántos registros fueron eliminados
duplicados_eliminados = registros_antes - registros_despues

# Mostrar los resultados
print("Registros antes de eliminar duplicados:", registros_antes)
print("Registros después de eliminar duplicados:", registros_despues)
print("Registros duplicados eliminados:", duplicados_eliminados)
```

**Tratamiento de cancelaciones y devoluciones**

Durante el análisis inicial se identificaron 10.624 registros con cantidades negativas.

La revisión realizada mostró que 9.288 de estos registros tienen un número de factura que comienza con la letra `C`, característica utilizada en este conjunto de datos para identificar cancelaciones.

Para este proceso ETL se busca construir un conjunto de datos orientado al análisis de las ventas realizadas. Por esta razón, las transacciones identificadas como cancelaciones serán excluidas del conjunto de datos final.

También se excluirán los registros con cantidades negativas que no presentan una factura iniciada en `C`, ya que durante el análisis se encontró que estos registros presentan características diferentes a una venta normal, como precio unitario igual a cero y ausencia de información del cliente.

Antes de aplicar la eliminación se calculará la cantidad de registros que serán afectados para poder verificar posteriormente el resultado.

***

```python
# Identificar las facturas que comienzan con la letra C
# Estas corresponden a cancelaciones dentro del conjunto de datos
es_cancelacion = (
    df_retail["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
)

# Identificar los registros que tienen cantidad negativa
es_cantidad_negativa = df_retail["Quantity"] < 0

# Contar cuántos registros cumplen cada condición
print("Registros con factura C:", es_cancelacion.sum())
print("Registros con cantidad negativa:", es_cantidad_negativa.sum())

# Contar cuántos registros cumplen al menos una de las dos condiciones
registros_a_excluir = es_cancelacion | es_cantidad_negativa

print("Registros que cumplen alguna de las condiciones:",
      registros_a_excluir.sum())
```

**Eliminación de cancelaciones y registros con cantidades negativas**

Después de analizar los registros con cantidades negativas, se determinó que todos los registros correspondientes a facturas iniciadas en `C` también presentan cantidades negativas.

Por lo tanto, para preparar un conjunto de datos orientado al análisis de ventas, se utilizará `Quantity < 0` como criterio para excluir las cancelaciones, devoluciones y demás registros con cantidades negativas.

Esta transformación permitirá trabajar posteriormente con las transacciones que presentan cantidades positivas.

***

```python
# Guardar la cantidad de registros antes de eliminar las cantidades negativas
registros_antes_negativos = len(df_retail)

# Conservar únicamente los registros con cantidades positivas
df_retail = df_retail[df_retail["Quantity"] > 0].copy()

# Guardar la cantidad de registros después de la transformación
registros_despues_negativos = len(df_retail)

# Calcular cuántos registros fueron eliminados
registros_negativos_eliminados = (
    registros_antes_negativos - registros_despues_negativos
)

# Mostrar los resultados
print(
    "Registros antes de eliminar cantidades negativas:",
    registros_antes_negativos
)

print(
    "Registros después de la transformación:",
    registros_despues_negativos
)

print(
    "Registros eliminados:",
    registros_negativos_eliminados
)
```

**Tratamiento de precios negativos**

Durante el análisis inicial se identificaron únicamente 2 registros con precios negativos.

Al revisar estos registros se encontró que ambos tienen la descripción `Adjust bad debt` y corresponden a ajustes relacionados con deuda, por lo que no representan ventas de productos.

Por esta razón, estos 2 registros serán excluidos del conjunto de datos utilizado para el análisis de ventas.

Los registros con `UnitPrice = 0` no serán eliminados en esta etapa, debido a que el análisis mostró que algunos corresponden a productos con información de transacción y podrían representar operaciones especiales.

***

```python
# Guardar la cantidad de registros antes de eliminar precios negativos
registros_antes_precio = len(df_retail)

# Conservar únicamente los registros cuyo precio sea mayor o igual a cero
df_retail = df_retail[df_retail["UnitPrice"] >= 0].copy()

# Guardar la cantidad de registros después de la transformación
registros_despues_precio = len(df_retail)

# Calcular cuántos registros fueron eliminados
precios_negativos_eliminados = (
    registros_antes_precio - registros_despues_precio
)

# Mostrar los resultados
print(
    "Registros antes de eliminar precios negativos:",
    registros_antes_precio
)

print(
    "Registros después de la transformación:",
    registros_despues_precio
)

print(
    "Precios negativos eliminados:",
    precios_negativos_eliminados
)
```

No se eliminarán estos registros porque la ausencia del identificador del cliente no significa que la transacción sea inválida y eliminar estas filas ocasionaría una pérdida importante de información para el análisis de ventas.

**Creación de la variable de valor total**

Se creará una nueva variable llamada `TotalVenta`.

Esta variable se calculará multiplicando la cantidad de productos vendidos (`Quantity`) por el precio unitario (`UnitPrice`).

La fórmula utilizada será:

TotalVenta = Quantity × UnitPrice

Esta nueva variable permitirá conocer el valor económico asociado a cada registro y posteriormente facilitará análisis como ventas por producto, país, cliente o periodo.

***

```python
# Calcular el valor total de cada registro de venta
df_retail["TotalVenta"] = (
    df_retail["Quantity"] * df_retail["UnitPrice"]
)

# Mostrar algunos registros para comprobar el cálculo
df_retail[["Quantity", "UnitPrice", "TotalVenta"]].head()
```

**Creación de variables relacionadas con la fecha**

La variable `InvoiceDate` contiene tanto la fecha como la hora de cada transacción.

Para facilitar futuros análisis empresariales, se crearán nuevas variables a partir de esta información. Estas variables permitirán analizar las ventas por año, mes y día sin tener que procesar nuevamente la fecha original.

Se crearán las variables `Año`, `Mes` y `Día`.

***

```python
# Extraer el año de cada transacción
df_retail["Año"] = df_retail["InvoiceDate"].dt.year

# Extraer el mes de cada transacción
df_retail["Mes"] = df_retail["InvoiceDate"].dt.month

# Extraer el día de cada transacción
df_retail["Día"] = df_retail["InvoiceDate"].dt.day

# Mostrar las variables creadas junto con la fecha original
df_retail[["InvoiceDate", "Año", "Mes", "Día"]].head()
```
